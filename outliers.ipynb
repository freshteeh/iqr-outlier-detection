```python
list=[1,2,3,4,5,6,7,8,9,30]
```

```python
min,Q1,median,Q3,max=np.quantile(list,[0.0,0.25,0.5,0.75,1.0])
```

```python
min,Q1,median,Q3,max
```

```python
IQR=Q3-Q1
print(IQR)
```

```python
lower_fence=Q1-1.5*IQR
upper_fence=Q3+1.5*IQR
```
